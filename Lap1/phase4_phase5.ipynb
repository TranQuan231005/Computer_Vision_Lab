```python
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open("cat.png")

print("Image loaded successfully!")
print("Size:", img.size)

plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
print(img)
```

## 4.1 Crop the Image

Select and crop a specific region of the original image.

**Description:**
The image is cropped by selecting a rectangular region using the coordinates `(100, 100, 400, 400)`. The cropped image is then displayed to show the result.


```python
# 4.1 - Crop the Image

crop_img = img.crop((100, 100, 400, 400))

plt.figure(figsize=(6, 6))
plt.imshow(crop_img)
plt.axis("off")
plt.title("Cropped Image")
plt.show()
```

## 4.2 Resize by Scale

Resize the image using a scale factor.

**Description:**
The original image is resized by multiplying its width and height by a scale factor of `0.5`.


```python
# 4.2 - Resize by Scale

scale = 0.5

new_width = int(img.width * scale)
new_height = int(img.height * scale)

resized_scale = img.resize((new_width, new_height))

plt.figure(figsize=(6, 6))
plt.imshow(resized_scale)
plt.axis("off")
plt.title("Resized Image by Scale")
plt.show()
```

## 4.3 Resize to a Fixed Size

Resize the image to a specific width and height.

**Description:**
The original image is resized to a fixed size of `400 × 300` pixels.


```python
# 4.3 - Resize to a Fixed Size

fixed_width = 400
fixed_height = 300

resized_fixed = img.resize((fixed_width, fixed_height))

plt.figure(figsize=(6, 5))
plt.imshow(resized_fixed)
plt.axis("off")
plt.title("Resized Image to Fixed Size")
plt.show()
```

## 4.4 Display Results

Display the original image and the processed images for comparison.

**Description:**
The original image, cropped image, scaled image, and fixed-size image are displayed together.


```python
# 4.4 - Display Results

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(crop_img)
plt.title("Cropped Image")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(resized_scale)
plt.title("Resized by Scale")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(resized_fixed)
plt.title("Fixed Size")
plt.axis("off")

plt.tight_layout()
plt.show()
```

## 5.1 Draw a Line

Draw a line on the image.

**Description:**
A line is drawn from the top-left area to the bottom-right area of the image.


```python
# 5.1 - Draw a Line

line_img = img.copy()

from PIL import ImageDraw

draw = ImageDraw.Draw(line_img)

draw.line((50, 50, 400, 300), fill="red", width=5)

plt.figure(figsize=(6, 6))
plt.imshow(line_img)
plt.axis("off")
plt.title("Draw a Line")
plt.show()
```

## 5.2 Draw a Circle

Draw a circle on the image.

**Description:**
A circle is drawn in the center area of the image using a bounding box.


```python
# 5.2 - Draw a Circle

circle_img = img.copy()

draw = ImageDraw.Draw(circle_img)

draw.ellipse((150, 100, 350, 300), outline="blue", width=5)

plt.figure(figsize=(6, 6))
plt.imshow(circle_img)
plt.axis("off")
plt.title("Draw a Circle")
plt.show()
```

## 5.3 Draw a Rectangle

Draw a rectangle on the image.

**Description:**
A rectangular shape is drawn around a selected region of the image.


```python
# 5.3 - Draw a Rectangle

rectangle_img = img.copy()

draw = ImageDraw.Draw(rectangle_img)

draw.rectangle((100, 100, 400, 300), outline="green", width=5)

plt.figure(figsize=(6, 6))
plt.imshow(rectangle_img)
plt.axis("off")
plt.title("Draw a Rectangle")
plt.show()
```

## 5.5 Display the Result

Display the final image with all drawing operations.

**Description:**
The final result combines a line, circle, rectangle, and text on the original image.


```python
# 5.5 - Display the Result

result_img = img.copy()

draw = ImageDraw.Draw(result_img)

# Draw line
draw.line((50, 50, 400, 300), fill="red", width=5)

# Draw circle
draw.ellipse((150, 100, 350, 300), outline="blue", width=5)

# Draw rectangle
draw.rectangle((100, 100, 400, 300), outline="green", width=5)

# Add text
draw.text((50, 350), "Computer Vision", fill="yellow")

plt.figure(figsize=(8, 6))
plt.imshow(result_img)
plt.axis("off")
plt.title("Final Drawing Result")
plt.show()
```
