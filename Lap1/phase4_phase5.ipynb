```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("pexels-conun-drum-37255800.jpg")

print(img.shape)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
crop = img[100:500, 100:500]

plt.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
resize_scale = cv2.resize(
    img,
    None,
    fx=0.5,
    fy=0.5
)

plt.imshow(cv2.cvtColor(resize_scale, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
resize_fixed = cv2.resize(
    img,
    (400,300)
)

plt.imshow(cv2.cvtColor(resize_fixed, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show() 
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap='gray')
plt.axis('off')
plt.show()
```

```python
blur = cv2.GaussianBlur(gray, (7,7), 0)

plt.imshow(blur, cmap='gray')
plt.axis('off')
plt.show()
```

```python
edges = cv2.Canny(blur, 100, 200)

plt.imshow(edges, cmap='gray')
plt.axis('off')
plt.show()
```

```python
print("Kích thước:", img.shape)
print("Chiều cao:", img.shape[0])
print("Chiều rộng:", img.shape[1])
```

```python
cv2.imwrite("edge_result.jpg", edges)
print("Đã lưu ảnh")
```

```python
crop = img[500:2000, 1000:3000]

plt.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
scale = 0.5

width = int(img.shape[1] * scale)
height = int(img.shape[0] * scale)

resized = cv2.resize(img, (width, height))

plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
fixed = cv2.resize(img, (800, 600))

plt.imshow(cv2.cvtColor(fixed, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
plt.title("Crop")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title("Resize Scale")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(cv2.cvtColor(fixed, cv2.COLOR_BGR2RGB))
plt.title("Fixed Size")
plt.axis('off')

plt.show()
```

```python
line_img = img.copy()

cv2.line(line_img,
         (100,100),
         (1000,1000),
         (255,0,0),
         5)

plt.imshow(cv2.cvtColor(line_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
circle_img = img.copy()

cv2.circle(circle_img,
           (1500,1500),  # tâm
           300,          # bán kính
           (0,255,0),    # màu xanh lá
           5)            # độ dày

plt.imshow(cv2.cvtColor(circle_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
rect_img = img.copy()

cv2.rectangle(rect_img,
              (500,500),     # góc trên trái
              (2000,1500),   # góc dưới phải
              (255,0,0),     # màu xanh dương
              5)

plt.imshow(cv2.cvtColor(rect_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
text_img = img.copy()

cv2.putText(text_img,
            "Ngoc Nhan",
            (500,500),
            cv2.FONT_HERSHEY_SIMPLEX,
            3,
            (255,0,0),
            5)

plt.imshow(cv2.cvtColor(text_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(20,10))

plt.subplot(2,2,1)
plt.imshow(cv2.cvtColor(line_img, cv2.COLOR_BGR2RGB))
plt.title("Line")
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(cv2.cvtColor(circle_img, cv2.COLOR_BGR2RGB))
plt.title("Circle")
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(cv2.cvtColor(rect_img, cv2.COLOR_BGR2RGB))
plt.title("Rectangle")
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(cv2.cvtColor(text_img, cv2.COLOR_BGR2RGB))
plt.title("Text")
plt.axis('off')

plt.show()
```
