### Task 2.2 - Display the Image

Display the original image using Matplotlib.
OpenCV uses BGR color order, so the image is converted to RGB before displaying.

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

### Task 2.3 - Save the Image

Convert the JPG image to PNG format and save it in the output directory using Pillow.

```python
pil_image = Image.fromarray(image_rgb)

output_path = "images/output/converted_image.png"

pil_image.save(output_path)

print("Image saved successfully!")
print("Output:", output_path)
```

## Phase 3 - Color Space Conversion

### Task 3.1 - Convert to Grayscale

Convert the original image from BGR color space to grayscale.

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

cv2.imwrite("images/output/grayscale.jpg", gray)

print("Grayscale conversion completed.")
```

### Task 3.2 - Convert to HSV

Convert the original image from BGR color space to HSV color space.

```python
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

cv2.imwrite("images/output/hsv.jpg", hsv)

print("HSV conversion completed.")
```

### Task 3.3 - Convert to LAB

Convert the original image from BGR color space to LAB color space.

```python
lab = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)

cv2.imwrite("images/output/lab.jpg", lab)

print("LAB conversion completed.")
```

### Task 3.4 - Display Results

Display the grayscale, HSV, and LAB conversion results.

```python
plt.figure(figsize=(15, 10))

# Grayscale
plt.subplot(2, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

# HSV
plt.subplot(2, 2, 2)
plt.imshow(hsv)
plt.title("HSV")
plt.axis("off")

# LAB
plt.subplot(2, 2, 3)
plt.imshow(lab)
plt.title("LAB")
plt.axis("off")

plt.tight_layout()
plt.show()
```
